# Download STL-10

**STL-10** is an image dataset from Stanford, built for learning features from
*unlabelled* images. It has only 5,000 labelled training images, but 100,000 images
without any label.

This notebook downloads it and unpacks it

Dataset: <https://cs.stanford.edu/~acoates/stl10/>

```python
import tarfile
import time
from pathlib import Path
from urllib.request import urlopen

DATA = Path("data")
DATA.mkdir(exist_ok=True)

URL = "https://cs.stanford.edu/~acoates/stl10/stl10_binary.tar.gz"
ARCHIVE = DATA / "stl10_binary.tar.gz"
EXTRACTED = DATA / "stl10_binary"

print("archive :", ARCHIVE)
print("unpacked:", EXTRACTED)
```

### Download

The binary archive is about **2.6 GB**. It is skipped if it is already there.

Progress is printed roughly every 256 MB.

```python
if ARCHIVE.exists():
    print(f"already downloaded ({ARCHIVE.stat().st_size / 1e9:.2f} GB) — skipping")
else:
    print(f"downloading {URL}")
    t0 = time.perf_counter()

    with urlopen(URL) as response, open(ARCHIVE, "wb") as f:
        total = int(response.headers.get("Content-Length", 0))
        done = 0
        next_report = 0

        while chunk := response.read(1 << 20):
            f.write(chunk)
            done += len(chunk)
            if done >= next_report:
                pct = 100 * done / total if total else float("nan")
                print(f"  {done / 1e9:5.2f} / {total / 1e9:.2f} GB  ({pct:5.1f}%)", flush=True)
                next_report = done + (256 << 20)

    print(f"done in {time.perf_counter() - t0:.0f} s")
```

### Unpack

It is a plain `.tar.gz`, so one call does it.

```python
if EXTRACTED.exists():
    print("already extracted thus skipping")
else:
    print("extracting ...")
    t0 = time.perf_counter()
    with tarfile.open(ARCHIVE, "r:gz") as tar:
        tar.extractall(DATA)
    print(f"done in {time.perf_counter() - t0:.0f} s")
```

### Data folder content

```python
total = 0
for p in sorted(EXTRACTED.iterdir()):
    size = p.stat().st_size
    total += size
    print(f"{p.name:20} {size / 1e6:9.1f} MB")

print(f"{'total':20} {total / 1e9:9.2f} GB")
```

Two files are much bigger than the rest: `train_X.bin` and especially `unlabeled_X.bin`.
