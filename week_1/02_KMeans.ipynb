# k-means

k-means repeats two steps until nothing changes:

1. assign every point to its nearest centre,
2. move every centre to the **mean** of the points assigned to it.

It minimises the *inertia*: the sum of squared distances from each point to its centre.

Two consequences follow directly from that definition, and we will watch both happen:

* a centre is a mean, so it is generally **not** one of the data points,
* a mean is pulled by outliers.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

d = np.load("data/blobs.npz")
X_clean, y_clean = d["X_clean"], d["y_clean"]
X_out = d["X_out"]

RANDOM_STATE = 0
K = 4
```

```python
def plot_kmeans(X, km, title):
    fig, ax = plt.subplots(figsize=(6, 5))
    ax.scatter(X[:, 0], X[:, 1], s=25, c=km.labels_, cmap="tab10",
               edgecolor="white", linewidth=0.5)
    ax.scatter(km.cluster_centers_[:, 0], km.cluster_centers_[:, 1],
               s=220, marker="X", c="black", label="centroids")
    ax.legend()
    ax.set_title(title)
    plt.tight_layout()
    plt.show()
```

### Clean blobs

```python
km_clean = KMeans(n_clusters=K, n_init=10, random_state=RANDOM_STATE).fit(X_clean)
plot_kmeans(X_clean, km_clean, f"k-means, clean blobs — inertia = {km_clean.inertia_:.1f}")
```

### With the outliers

Watch the centroids. A mean has to sit *between* the points it averages, so a centroid
that shares a cluster with outliers gets dragged out into empty space and takes the
cluster boundary with it.

```python
km_out = KMeans(n_clusters=K, n_init=10, random_state=RANDOM_STATE).fit(X_out)
plot_kmeans(X_out, km_out, f"k-means, with outliers — inertia = {km_out.inertia_:.1f}")
```
