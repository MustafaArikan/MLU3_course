# Check the environment

Run this first. It confirms that the `uv` environment for `week_1` is active and that
the packages we need are importable.

```python
import sys
import numpy as np
import pandas as pd
import matplotlib as mpl
import sklearn

print(f"Python:     {sys.version.split()[0]}")
print(f"Executable: {sys.executable}")
print("-" * 40)

packages = {
    "NumPy": np.__version__,
    "pandas": pd.__version__,
    "scikit-learn": sklearn.__version__,
    "Matplotlib": mpl.__version__,
}

for name, version in packages.items():
    print(f"{name.ljust(15)}: {version}")
```

If an import fails, run `uv sync` in this folder and make sure the **week-1** kernel is
selected
